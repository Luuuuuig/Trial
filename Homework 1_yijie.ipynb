Let $X \in \mathbb{R}^{m\times p}$ be a centered data matrix where each row
represents an observation and each column represents a feature. The centering
condition means each column has zero mean: $\sum_{i=1}^m X_{i,j} = 0$ for all $j$.

## 1a. Eigenvalues and Total Variance
Given the eigendecomposition of the sample covariance matrix
$$
C = \frac{1}{m-1} X^\top X = Q \Lambda Q^\top
$$
where $Q$ contains the eigenvectors and $\Lambda$ is diagonal with eigenvalues
$\lambda_1, \lambda_2, \ldots, \lambda_p$, the total variance in the dataset equals:
$$
\text{Total Sample Variance} = \operatorname{tr}(C) = \frac{1}{m-1}\operatorname{tr}(X^\top X) = \lambda_{max}
$$

○ False
○ True

**Answer:** False – the total sample variance is $\operatorname{tr}(C)=\sum_{j=1}^p \lambda_j$, i.e. the sum of all eigenvalues, not just the largest eigenvalue $\lambda_{max}$.  

## 1b. Covariance for Centered Data
For a centered data matrix $ X $, the sample covariance between features $ j $ and $ k $ can be computed as:

$$
\operatorname{Cov}(j,k) = \frac{1}{m-1} \, X_{(:,j)}^\top X_{(:,k)}
$$

where $ X_{(:,j)} $ denotes the $ j $-th column of $ X $.

○ False
○ True

**Answer:** True – with centered features this dot product scaled by $1/(m-1)$ is exactly the sample covariance between columns $j$ and $k$.  

## 1c. Covariance Under Linear Transformations

If we apply a linear transformation

$$
Y = X W
$$

where $ W \in \mathbb{R}^{p\times q} $ is a transformation matrix, then the covariance matrix of the transformed features is:

$$
C_Y = W^\top C_X W
$$

where

$$
C_X = \frac{X^\top X}{m-1}
$$

is the covariance matrix of $ X $.

○ False
○ True

**Answer:** True – substituting $Y=XW$ into $C_Y=\tfrac{1}{m-1}Y^\top Y$ yields $W^\top C_X W$.  

## 1d. Matrix Norms and Trace Properties
For any matrix $ A \in \mathbb{R}^{n\times d} $, the following relationship holds between row-wise and column-wise squared norms:

$$
\operatorname{tr}(A A^\top) = \sum_{i=1}^n \|A_{(i,:)}\|^2 = \sum_{j=1}^d \|A_{(:,j)}\|^2
$$

○ False
○ True

**Answer:** True – $\operatorname{tr}(AA^\top)$ adds the squared row norms and, by the cyclic property of the trace, also equals the sum of squared column norms.

Raw data often comes with features on vastly different scales. Imagine comparing house prices in dollars with number of bedrooms, or pixel intensities (0-255) with age in years. Without proper scaling, algorithms can be biased toward features with larger numerical ranges, leading to poor performance.

Data normalization addresses this by transforming features to comparable scales while preserving the underlying relationships in the data. Different scaling methods make different assumptions about the data distribution and have varying robustness to outliers:

- **Min-Max Scaling** transforms features to a fixed range (typically [0,1]), preserving the original distribution shape but being sensitive to outliers
- **Standardization (Z-score)** centers data around zero with unit variance, assuming normally distributed data
- **Robust Scaling** uses median and interquartile range, making it less sensitive to extreme values

````python
## 2a–2c. Scaling utilities and results
import numpy as np
import pandas as pd

sales_df = pd.DataFrame({
    "F1": [50, 70, 65, 500, 60],
    "F2": [2000, 2500, 2100, 20000, 2200],
    "F3": [5, 7, 6, 40, 5]
})


def minmax_scale(series: pd.Series) -> pd.Series:
    values = series.to_numpy(dtype=float)
    scaled = (values - values.min()) / (values.max() - values.min())
    return pd.Series(np.round(scaled, 2), index=series.index)


def standardize(series: pd.Series) -> pd.Series:
    values = series.to_numpy(dtype=float)
    scaled = (values - values.mean()) / values.std(ddof=1)
    return pd.Series(np.round(scaled, 2), index=series.index)


def robust_scale(series: pd.Series) -> pd.Series:
    values = series.to_numpy(dtype=float)
    median = float(np.median(values))
    sorted_vals = np.sort(values)
    lower = sorted_vals[: len(sorted_vals) // 2]
    upper = sorted_vals[len(sorted_vals) // 2 + 1 :]
    q1 = float(np.median(lower))
    q3 = float(np.median(upper))
    iqr = q3 - q1
    scaled = (values - median) / iqr
    return pd.Series(np.round(scaled, 2), index=series.index)

# Compute requested transformations
f1_minmax = minmax_scale(sales_df["F1"])
f2_standardized = standardize(sales_df["F2"])
f3_robust = robust_scale(sales_df["F3"])

print("Min-max scaled F1:", f1_minmax.tolist())
print("Standardized F2 at ID=4:", f2_standardized.loc[4])
print("Robust scaled F3 at ID=3:", f3_robust.loc[3])

summary = pd.DataFrame({
    "Original_F1": sales_df["F1"],
    "F1_minmax": f1_minmax,
    "Original_F2": sales_df["F2"],
    "F2_standardized": f2_standardized,
    "Original_F3": sales_df["F3"],
    "F3_robust": f3_robust,
})
summary
````

The transformations make the feature scales directly comparable. Min-max scaling stretches **F1** into $[0,1]$ while retaining the relative ordering, standardization centers **F2** around zero with unit variance so the revenue outlier at ID 3 remains evident, and robust scaling keeps **F3** resistant to the extreme return count (ID 3) by using the Tukey interquartile range (yielding a much smaller scaled value than z-score scaling would).

# 3. Gradient Descent

In this exercise, you'll implement and explore gradient descent, a cornerstone of modern optimization. The method is simple in idea but powerful in application: we iteratively update our variables in the direction that most rapidly decreases the function, based on the **negative gradient**. While coordinate descent updates one variable at a time, gradient descent moves in the direction of steepest descent across all coordinates simultaneously. This often leads to:

- Faster convergence when variables are tightly coupled or when the function isn’t naturally separable,
- More direct progress toward a minimum, especially in smooth, well-behaved functions,
- Better behavior in high-dimensional settings where dependencies across variables are strong.

````python
import math
import numpy as np
import pandas as pd

# Himmelblau's function and its gradient
def f(u, v):
    A = u**2 + v - 11
    B = u + v**2 - 7
    return A * A + B * B


def grad_f(u, v):
    A = u**2 + v - 11
    B = u + v**2 - 7
    du = 4 * u * A + 2 * B
    dv = 2 * A + 4 * v * B
    return du, dv


def eta_const(t, c=1e-3):
    return c


def eta_sqrt(t, c=1e-3):
    return c / math.sqrt(t + 1)


def eta_multistep(t, milestones=[20, 50], c=0.1, eta_init=1e-3):
    eta = eta_init
    for m in milestones:
        if t >= m:
            eta *= c
    return eta


def gradient_descent(f, grad_f, eta, u0, v0, max_iter=100):
    u, v = float(u0), float(v0)
    path = [(u, v)]
    values = [f(u, v)]
    for t in range(max_iter):
        du, dv = grad_f(u, v)
        step = eta(t)
        u -= step * du
        v -= step * dv
        path.append((u, v))
        values.append(f(u, v))
    return path, values


def summarize_run(path, values):
    u_final, v_final = path[-1]
    return {
        "u100": round(u_final, 6),
        "v100": round(v_final, 6),
        "f(u100,v100)": round(values[-1], 6),
        "best_f_1..100": round(min(values[1:]), 6),
    }

# Experiments requested in 3a–3c
strategies = {
    "const (c=1e-3)": lambda t: eta_const(t, 1e-3),
    "sqrt decay (c=1e-3)": lambda t: eta_sqrt(t, 1e-3),
    "multi-step [20,50]": lambda t: eta_multistep(t, [20, 50], 0.1, 1e-3),
}

strategy_rows = []
for name, schedule in strategies.items():
    path, values = gradient_descent(f, grad_f, schedule, 4, -5, max_iter=100)
    row = summarize_run(path, values)
    row.update({"strategy": name})
    strategy_rows.append(row)

df_strats = pd.DataFrame(strategy_rows)[
    ["strategy", "u100", "v100", "f(u100,v100)", "best_f_1..100"]
]

# Experiments for 3d with different initializations using eta_const
initial_points = [(-4, 0), (0, 0), (4, 0), (0, 4), (5, 5)]
init_rows = []
for (u0, v0) in initial_points:
    path, values = gradient_descent(f, grad_f, lambda t: eta_const(t, 1e-3), u0, v0, max_iter=100)
    row = summarize_run(path, values)
    row.update({"init": f"({u0}, {v0})"})
    init_rows.append(row)

df_inits = pd.DataFrame(init_rows)[
    ["init", "u100", "v100", "f(u100,v100)", "best_f_1..100"]
]
````

# 4. Coordinate Descent

In optimization, we often aim to find the global minimum of a function by solving the gradient equation analytically. However, this becomes impractical in many real-world settings. Functions may be too complex, too high-dimensional, or lack closed-form solutions. This is where **coordinate descent** provides an alternative.

Coordinate descent updates one variable at a time, minimizing along each coordinate direction while holding the others fixed. This is particularly useful when:
- The function is differentiable but hard to minimize jointly,
- Partial updates are much easier to compute (analytically or numerically).

In this assignment, you'll apply coordinate descent to the following function:

$$
f(\mathbf{x}) = \exp(x_1 - 3x_2 + 3) + \exp(3x_2 - 2x_3 - 2) + \exp(2x_3 - x_1 + 2)
$$
with $x_1, x_2, x_3 \in \mathbb{R}$.

This function is fully differentiable, and its coordinate-wise argmins are analytically tractable. In part $a$, you’ll derive the updates for each coordinate individually. In part $b$, you’ll use these updates to implement a full coordinate descent loop, stepping through each coordinate iteratively and observing the convergence behavior.

````python
import math


def f(x1, x2, x3):
    return math.exp(x1 - 3 * x2 + 3) + math.exp(3 * x2 - 2 * x3 - 2) + math.exp(2 * x3 - x1 + 2)


def argmin_x1(x1, x2, x3):
    return x3 + 1.5 * x2 - 0.5


def argmin_x2(x1, x2, x3):
    return (x1 + 2 * x3 + 5) / 6.0


def argmin_x3(x1, x2, x3):
    return (x1 + 3 * x2 - 4) / 4.0


def coordinate_descent(x0, max_iter=25):
    x1, x2, x3 = x0
    history = []
    for _ in range(max_iter):
        x1 = argmin_x1(x1, x2, x3)
        x2 = argmin_x2(x1, x2, x3)
        x3 = argmin_x3(x1, x2, x3)
        history.append((x1, x2, x3, f(x1, x2, x3)))
    return history

x0 = (1.0, 20.0, 5.0)
first_update = (
    argmin_x1(*x0),
    argmin_x2(*x0),
    argmin_x3(*x0),
)
final_state = coordinate_descent(x0, max_iter=25)[-1]

print("Single-coordinate minimizers at t0=(1,20,5):")
print(f"  argmin_x1 = {first_update[0]:.4f}")
print(f"  argmin_x2 = {first_update[1]:.4f}")
print(f"  argmin_x3 = {first_update[2]:.4f}")

print("\nAfter 25 coordinate descent iterations starting from (1,20,5):")
print(f"  x1 = {final_state[0]:.4f}")
print(f"  x2 = {final_state[1]:.4f}")
print(f"  x3 = {final_state[2]:.4f}")
print(f"  f(x) = {final_state[3]:.4f}")
````

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Prepare training dataframe for feature analysis
train_df = pd.DataFrame(X_train, columns=feature_names)
train_df['MedHouseVal'] = y_train

variance = train_df[feature_names].var()
corr_with_target = train_df.corr(numeric_only=True)['MedHouseVal'].drop('MedHouseVal')
corr_matrix = train_df[feature_names].corr()

candidates = ['AveBedrms', 'Population', 'AveOccup']
feature_stats = pd.DataFrame({
    'Variance': variance[candidates],
    'Corr_with_target': corr_with_target[candidates],
    'Corr_with_AveRooms': [corr_matrix.loc['AveRooms', feat] for feat in candidates],
}).round(4)

print("Candidate feature statistics (training split):")
print(feature_stats)

kf = KFold(n_splits=5, shuffle=True, random_state=42)

def evaluate_subset(feature_subset):
    idx = [feature_names.index(f) for f in feature_subset]
    pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('regressor', LinearRegression())
    ])
    cv_scores = cross_val_score(pipeline, X_train[:, idx], y_train, scoring='neg_mean_squared_error', cv=kf)
    pipeline.fit(X_train[:, idx], y_train)
    val_mse = mean_squared_error(y_val, pipeline.predict(X_val[:, idx]))
    return -cv_scores.mean(), cv_scores.std(), val_mse

models = {
    'All features': list(feature_names),
}
for feat in candidates:
    models[f'Without {feat}'] = [f for f in feature_names if f != feat]

rows = []
for model_name, feats in models.items():
    mean_cv, std_cv, val_mse = evaluate_subset(feats)
    rows.append({
        'Model': model_name,
        'Num features': len(feats),
        'CV MSE (mean)': round(mean_cv, 3),
        'CV MSE (std)': round(std_cv, 3),
        'Validation MSE': round(val_mse, 3),
    })

cv_results = pd.DataFrame(rows)
print()
print("Cross-validation summary (5-fold, Ridge-free linear regression):")
print(cv_results.to_string(index=False))
````
